```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Carga desde un archivo .csv sin indice
data= pd.read_csv('Base_de_Leads_Sin_Nulos.csv', index_col=False) 
data.head()
```

```python
#Verificamos información del DataFrame
data.info()
```

```python
# Corroboramos valores nulos
valores_nulos=data.isnull().sum()
valores_nulos
```

```python
# Creo 2 dataframe para poder procesar los outliers
cuantitativas= data[['Mes Lead','Unnamed: 0']]
cualitativas= data[['Fecha de Alta', 'AgLead','Titulo','Nombre','Producto','Temperatura','Comentario','Asesor','Fuente','Origen']]
```

```python
# Realizamos diagrama de caja o bigote de cada columna del dataframe
fig = plt.figure(figsize =(15, 8))
cuantitativas.plot(kind='box', vert=False)
plt.title("Valores Atípicos del Dataframe")
plt.show() #dibujamos el diagrama
```

```python
# Unimos el dataframe cuantitativo limpio con el dataframe cualitativo
Datos_limpios = pd.concat([cualitativas, cuantitativas], axis=1)
Datos_limpios
```

```python
# Convertir DataFrame a CSV
Datos_limpios.to_csv("Base_de_leads_sin_outliers.csv")
```
