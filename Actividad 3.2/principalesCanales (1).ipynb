```python
#Importamos las librerias pandas, numpy y matplotlib respectivamente
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Cargamos el archivo
data=pd.read_csv("Principales_Canales_Limpio.csv")

#Mostramos los primeros registros
data.head()
```

```python
#Verificamos información del DataFrame
data.info()
```

```python
#Verificamos los valores nulos
valores_nulos=data.isnull().sum()
valores_nulos
```

```python
#Revisamos los tipos de datos de cada columna
data.dtypes
```

```python
#Convertimos datos a tipo numerico
for columna in data.columns[1:31]:
    data[columna] = pd.to_numeric(data[columna], errors='coerce')

data.head()
```

```python
#Seleccionamos los datos cuantitativos de Piso
cuantitativas = data.iloc[5:14, 1:31]

#Cambiamos filas por columnas
cuantitativas = cuantitativas.T

cuantitativas
```

```python
#Ponemos nombres a las columnas
cuantitativas.columns = data.iloc[5:14, 0]

#Reemplazamos valores nulos por 0
cuantitativas = cuantitativas.fillna(0)

cuantitativas
```

```python
#Realizamos diagrama de caja o bigote de cada columna del dataframe

fig = plt.figure(figsize =(15, 8))
cuantitativas.plot(kind='box', vert=False)
plt.title("Valores Atípicos del Dataframe")
plt.show() 
```

```python
#Método aplicando desviación estandar. Encuentro los valores extremos

y=cuantitativas
Limite_Superior= y.mean() + 3*y.std()
Limite_Inferior= y.mean() - 3*y.std()
print("Limite superior permitido", Limite_Superior)
print("Limite inferior permitido", Limite_Inferior)
```

```python
#Obtenemos datos y los outliers se convierten en nulos en el DataFrame

data3= cuantitativas[(y<=Limite_Superior)&(y>=Limite_Inferior)]
data3
```

```python
#Corroboramos valores nulos del dataframe

valores_nulos=data3.isnull().sum()
valores_nulos
```

```python
#Método aplicando Cuartiles. Encuentro cuartiles 0.25 y 0.75
y=cuantitativas

percentile25=y.quantile(0.25) #Q1
percentile75=y.quantile(0.75) #Q3
iqr= percentile75 - percentile25

Limite_Superior_iqr= percentile75 + 1.5*iqr
Limite_Inferior_iqr= percentile25 - 1.5*iqr
print("Limite superior permitido", Limite_Superior_iqr)
print("Limite inferior permitido", Limite_Inferior_iqr)
```

```python
#Obtenemos datos limpios del Dataframe

data3_iqr= cuantitativas[(y<=Limite_Superior_iqr)&(y>=Limite_Inferior_iqr)]

#Corroboramos valores nulos del dataframe LIMPIO

valores_nulos=data3_iqr.isnull().sum()
valores_nulos
```

```python
#Reemplazamos valores atípicos (nulos) del dataframe con "mean"

#Realizamos una copia del dataframe

data4_iqr=data3_iqr.copy()
data4_iqr=data4_iqr.fillna(round(data3_iqr.median(),1))
data4_iqr
```

```python
#Corroboramos valores nulos del dataframe LIMPIO

valores_nulos=data4_iqr.isnull().sum()
valores_nulos
```
